# Reproducing every figure in the paper

This notebook recomputes the figures of the paper.

We only use a curated subset of the raw data, which is included in this repository. The full raw data is not included due to its size and privacy concerns.

**The curated subset reproduces the full pipeline's output bit-for-bit**: the plot scripts produce identical figures from the full raw tree and from this copy.


```python
import sys, os, subprocess
from IPython.display import Image, display, Markdown

PLOTTING_DIR = os.path.join(os.getcwd(), "plotting")
sys.path.insert(0, PLOTTING_DIR)
os.chdir(PLOTTING_DIR)

print("Working from:", os.getcwd())
```

## Step 0 Base-map terrain for Figure 3

Figure 3 uses a hillshade built from open terrain tiles (Mapzen Terrain Tiles
on AWS; Europe: Copernicus EU-DEM). The tiles are not shipped with the
repository, so they are downloaded once here if missing (needs internet; the
OpenStreetMap and Natural Earth base-map files are already included).

```python
BASEMAP_DEM = os.path.join(os.path.dirname(PLOTTING_DIR), "data", "study_area_basemap", "dem", "index.json")
if not os.path.exists(BASEMAP_DEM):
    r = subprocess.run([sys.executable, "fetch_study_area_basemap.py"], capture_output=True, text=True)
    print(r.stdout[-2000:])
    if r.returncode != 0:
        print(r.stderr[-2000:])
        raise RuntimeError("fetch_study_area_basemap.py failed (internet access needed)")
else:
    print("Terrain tiles already present.")
```
